Polynomial Regression + Gradient/
Degree = 2

```python
import numpy as np

# Data
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 6, 11, 18, 27], dtype=float)

# Polynomial features
x2 = x ** 2

# Initial weights
w0 = 0.0
w1 = 0.0
w2 = 0.0

# Hyperparameters
learning_rate = 0.001
epochs = 2000

n = len(x)

for epoch in range(epochs):

    # -------------------------
    # 1. Prediction
    # -------------------------
    y_pred = w0 + w1 * x + w2 * x2

    # -------------------------
    # 2. Error
    # -------------------------
    error = y_pred - y

    # -------------------------
    # 3. Loss (MSE)
    # -------------------------
    loss = np.mean(error ** 2)

    # -------------------------
    # 4. Gradients
    # -------------------------
    dw0 = np.mean(error)
    dw1 = np.mean(error * x)
    dw2 = np.mean(error * x2)

    # -------------------------
    # 5. Weight update
    # -------------------------
    w0 = w0 - learning_rate * dw0
    w1 = w1 - learning_rate * dw1
    w2 = w2 - learning_rate * dw2

    # Print progress
    if epoch % 100 == 0:
        print(
            f"Epoch: {epoch}, "
            f"Loss: {loss:.4f}, "
            f"w0: {w0:.4f}, "
            f"w1: {w1:.4f}, "
            f"w2: {w2:.4f}"
        )

print("\nFinal Parameters:")
print("w0 =", w0)
print("w1 =", w1)
print("w2 =", w2)
```

Polynomial Regression + Gradient/
Degree = 3

```python
import numpy as np

# =========================
# 1. Data
# =========================

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([6, 18, 48, 102, 186], dtype=float)


# =========================
# 2. Polynomial Features
# =========================

x1 = x
x2 = x ** 2
x3 = x ** 3


# =========================
# 3. Initial Weights
# =========================

w0 = 0.0
w1 = 0.0
w2 = 0.0
w3 = 0.0


# =========================
# 4. Hyperparameters
# =========================

learning_rate = 0.0001
epochs = 20000

n = len(x)


# =========================
# 5. Gradient Descent
# =========================

for epoch in range(epochs):

    # -------------------------
    # Prediction
    # -------------------------

    y_pred = (
        w0
        + w1 * x1
        + w2 * x2
        + w3 * x3
    )


    # -------------------------
    # Error
    # -------------------------

    error = y_pred - y


    # -------------------------
    # Loss
    # -------------------------

    loss = np.mean(error ** 2)


    # -------------------------
    # Gradients
    # -------------------------

    dw0 = np.mean(error)

    dw1 = np.mean(error * x1)

    dw2 = np.mean(error * x2)

    dw3 = np.mean(error * x3)


    # -------------------------
    # Weight Update
    # -------------------------

    w0 = w0 - learning_rate * dw0

    w1 = w1 - learning_rate * dw1

    w2 = w2 - learning_rate * dw2

    w3 = w3 - learning_rate * dw3


    # -------------------------
    # Show Progress
    # -------------------------

    if epoch % 1000 == 0:

        print(
            f"Epoch: {epoch} | "
            f"Loss: {loss:.4f} | "
            f"w0: {w0:.4f} | "
            f"w1: {w1:.4f} | "
            f"w2: {w2:.4f} | "
            f"w3: {w3:.4f}"
        )


# =========================
# 6. Final Parameters
# =========================

print("\nFinal Parameters:")

print("w0 =", w0)
print("w1 =", w1)
print("w2 =", w2)
print("w3 =", w3)


# =========================
# 7. Final Predictions
# =========================

y_pred = (
    w0
    + w1 * x
    + w2 * x**2
    + w3 * x**3
)

print("\nActual y:")
print(y)

print("\nPredicted y:")
print(y_pred)
```
