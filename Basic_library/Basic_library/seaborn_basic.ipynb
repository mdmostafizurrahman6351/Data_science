```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.scatter(x,y)
sns.set_theme() # seaborn style plot
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.scatter(x,y)
sns.set_theme(style="whitegrid") # seaborn style plot
plt.show()
```
