
linestyle="-"=Solid,
linestyle="--"=Dashed,
linestyle=":"=Dotted,
linestyle="-."=Dash-dot

```python
z = [0,2,4,6]
plt.plot(
    x,
    y,
    color='y', # line color
    linewidth=4, # line width
    linestyle='-.'
)
plt.plot(
    z,
    y,
    color='g', # color green
    lw=4, # line width
    linestyle='--' # line style

)
plt.show()
```


"o"=○ Circle,
"s"=□ Square ,
"^"=▲ Triangle,
"*"=★ Star,
"x"=✕ Cross,
"+"=+ Plus,
"D"=◇ Diamond

```python
z = [0,2,4,6]
plt.plot(
    x,
    y,
    color='y', # line color
    linewidth=1, # line width
    linestyle='-.', # line style
    marker='^' # marker style
)
plt.plot(
    z,
    y,
    color='g', # color green
    lw=2, # line width
    linestyle='--', # line style
    marker='*' # marker style
)
plt.show()
```

```python
z = [0,2,4,6]
p = [0,2,3,4]
plt.plot(
    p,
    y,
    color='k', # line color
    linestyle=':', # line style
    marker='^', # marker style
    ms=20, # marker size
    mec='y', # marker edge color
    mfc='blue' # marker face color
)
plt.plot(
    x,
    y,
    color='red',
    linestyle='-.',
    marker='*',
    ms=25,
    mec='blue',
    mfc='y'
)
plt.plot(
    z,
    y,
    color='g',
    linestyle='--',
    marker='o',
    ms=10,
    mec='red',
    mfc='y'
)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.scatter(
    x,
    y,
    color="blue",
    s=120,
    alpha=0.6, # Transparency
    marker="o"
)

plt.title("Scatter Plot")
plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

x1 = [1, 2, 3]
y1 = [2, 3, 2]

x2 = [5, 6, 7]
y2 = [5, 6, 5]

plt.scatter(
    x1,
    y1,
    color="blue",
    marker="o",
    s=100,
    label="Class A"
)

plt.scatter(
    x2,
    y2,
    color="red",
    marker="^",
    s=100,
    label="Class B"
)

plt.title("Classification Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 5, 2, 6, 4]

labels = [0, 1, 2, 1, 0]

plt.scatter(
    x,
    y,
    c=labels,
    cmap="plasma",
    s=150
)

plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 6]

labels = [0, 1, 2, 3, 4]

plt.scatter(
    x,
    y,
    c=labels,
    cmap="plasma",
    s=120
)

plt.colorbar()

plt.title("Scatter Plot")
plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Apple", "Banana", "Orange"]

sales = [50, 80, 40]

plt.bar(
    products,
    sales,
    color="blue",
    width=0.5
)

plt.title("Fruit Sales")
plt.xlabel("Fruit")
plt.ylabel("Sales")
plt.grid(axis="y")

plt.show()
```

```python
import matplotlib.pyplot as plt

features = [
    "Age",
    "Salary",
    "Experience"
]

importance = [
    0.65,
    0.25,
    0.10
]

plt.barh(
    features,
    importance,
    color="orange"
)

plt.title("Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.grid(axis="x")

plt.show()
```

```python
import matplotlib.pyplot as plt

data = [
    2,3,3,4,5,
    5,5,6,7,8
]

plt.hist(
    data,
    bins=5,
    color="skyblue",
    edgecolor="black"
)

plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

data = [
    12,15,18,20,22,
    25,27,30,32,35,
    38,40,42,45
]

plt.hist(
    data,
    bins=4,
    color="skyblue",
    edgecolor="black"
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [
    45,50,55,60,65,
    70,70,75,80,85,
    90,95
]

plt.hist(
    marks,
    bins=5,
    density=True,
    color="skyblue",
    edgecolor="black"
)

plt.title("Probability Distribution")
plt.xlabel("Marks")
plt.ylabel("Density")

plt.show()
```

```python
import matplotlib.pyplot as plt

sizes = [50, 30, 20]

labels = [
    "Cat",
    "Dog",
    "Bird"
]

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    colors=[
        "orange",
        "skyblue",
        "lightgreen"
    ],
    explode=[0.1, 0, 0],
    shadow=True,
    startangle=90
)

plt.title("Animal Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,4))

ax.plot(
    [10,20,15,30],
    marker="o",
    color="blue"
)

ax.set_title("Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Amount")

ax.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2)

ax[0].plot([1,2,3], [2,4,3])
ax[0].set_title("Accuracy")

ax[1].plot([1,2,3], [5,3,4])
ax[1].set_title("Loss")

plt.show()
```
