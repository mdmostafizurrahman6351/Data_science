```python
a = np.arange(15)
print(a)
print(a.ndim)
print(a.shape)

print("\n")

a = np.reshape(a,(3,5))
print(a)
print(a.ndim)
print(a.shape)
```

```python
# multiplaction
a = np.arange(15).reshape(3,5)
b = np.arange(15).reshape(5,3)
result = np.dot(a,b)
print(result)
```

```python
print(a.sum())
print(a.min())
print(a.max())
print(a.mean())
print(a.std())
print(a.var())
```

```python
a = np.arange(1,10).reshape(3,3)
print(a)
print("")
print(a.sum(axis=0)) #colam sum
print("")
print(a.sum(axis=1)) # row sum
```

```python
a = np.arange(1,16,2)
b = np.arange(1,9,1)
print(a)
print(b)
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
print(a//b)
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
print(np.mod(a,b))
print(np.power(a,b))
print(np.floor_divide(a,b))
print(np.exp(a))
print(np.log(a))
print(np.sqrt(a))
```
