```python
student = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Imran", "Ali", "Rahim"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 78]
})

pd.merge(student, marks, on="ID") # "merage" to data frame add
```

```python
pd.concat([student, marks]) # "concat" to data frame add raw wayes
```

```python
pd.concat([student, marks], axis=1) # columns wayes add
```
