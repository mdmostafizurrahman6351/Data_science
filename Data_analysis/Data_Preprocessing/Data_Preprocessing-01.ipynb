```python
import pandas as pd
```

```python
data = {
    "Age": [25, 30, 35, 40, 28, 30],
    "Salary": [30000, 40000, None, 60000, 35000, 40000],
    "Gender": ["Male", "Female", "Male", "Female", None, "Female"],
    "City": ["Dhaka", "Dhaka", "Chittagong", "Dhaka", "Rajshahi", "Dhaka"]
}

df = pd.DataFrame(data)
```

***Topic 02 — Missing Values***

```python
df.isnull().sum()
```

```python
(df.isnull().mean() * 100 ).round(2)
```

```python
df['Salary'].fillna(df['Salary'].median(), inplace=True)
```

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
```
